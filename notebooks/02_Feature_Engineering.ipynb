```python
import pandas as pd
import numpy as np

# Load Processed data 
df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

print("Data loaded!")
print("Shape:", df.shape)
```

```python
# New Feature 1: Years Since Promotion
df['YearsSincePromotion'] = df['YearsAtCompany'] - df['YearsSinceLastPromotion']

# New Feature 2: Salary Per Job Level
df['SalaryPerLevel'] = df['MonthlyIncome'] / df['JobLevel']

# New Feature 3: Overwork Score
df['OverworkScore'] = (df['OverTime'] == 'Yes').astype(int) * df['YearsInCurrentRole']

print("New features created!")
print(df[['YearsSincePromotion', 'SalaryPerLevel', 'OverworkScore']].head())
```
