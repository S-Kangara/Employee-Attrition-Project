```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#loading data
df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

# check first 5 rows 
df.head()
```

```python
# check data size and columns 
print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# check the attrition rate 
print("Attrition Count:")
print(df['Attrition'].value_counts())

print("\nAttrition Percentage:")
print(df['Attrition'].value_counts(normalize=True) * 100)
```

```python
# Attrition chart 
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Attrition', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=False)
plt.title('Employee Attrition Count')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('../outputs/01_attrition_count.png')
plt.show()
```

```python
# Department wise attrition
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Department', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=True)
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('../outputs/02_department_attrition.png')
plt.show()
```

```python
# Overtime vs Attrition
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=True)
plt.title('Attrition by Overtime')
plt.xlabel('OverTime')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('../outputs/03_overtime_attrition.png')
plt.show()
```

```python
# Age vs Attrition
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Attrition', y='Age', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=False)
plt.title('Age vs Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.tight_layout()
plt.savefig('../outputs/04_age_attrition.png')
plt.show()
```

```python
# Monthly Income vs Attrition
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=False)
plt.title('Monthly Income vs Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.tight_layout()
plt.savefig('../outputs/05_income_attrition.png')
plt.show()
```

```python
# Job Satisfaction vs Attrition
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='JobSatisfaction', hue='Attrition', palette=['#2ecc71','#e74c3c'], legend=True)
plt.title('Job Satisfaction vs Attrition')
plt.xlabel('Job Satisfaction (1=Low, 4=High)')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('../outputs/06_jobsatisfaction_attrition.png')
plt.show()
```
