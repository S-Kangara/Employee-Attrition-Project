```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('C:/Users/Sathsarani Kangara/Desktop/employee-attrition-predictor/data/processed_data.csv')

print("Data loaded!")
print("Shape:", df.shape)
```

```python
# Top features that cause attrition
features = ['OverTime', 'MonthlyIncome', 'Age', 'YearsAtCompany', 'YearsInCurrentRole']
attrition_rates = []

for feature in features:
    rate = df[df['Attrition'] == 1][feature].mean()
    attrition_rates.append(rate)

# Chart
plt.figure(figsize=(10, 6))
bars = plt.bar(features, attrition_rates, color=['#e74c3c', '#3498db', '#2ecc71', '#f39c12', '#9b59b6'])
plt.title('Average Values for Employees Who Left', fontsize=14)
plt.xlabel('Features')
plt.ylabel('Average Value')
plt.tight_layout()
plt.savefig('C:/Users/Sathsarani Kangara/Desktop/employee-attrition-predictor/outputs/09_business_insights.png')
plt.show()
```
